```python
import numpy as np
import matplotlib.pyplot as plt

scale = 1
shape = 1

muestra_uniformes = np.random.uniform(0,1, 1000)
weibull = scale*(-np.log(1-muestra_uniformes))**(shape)
plt.hist(weibull, bins = 50, density = True, color = "b")

x = np.linspace(0, np.max(weibull), 100)
pdf = (shape*scale)*(x*scale)**(shape-1)*np.exp(-(scale*x)**shape)
plt.plot(x, pdf, "r-", label = "Distribución Weibull")

plt.xlabel("Valores")
plt.ylabel("Frecuencia")
plt.title("Distribución Weibull a partir de muestras uniformes")
plt.legend()
plt.show()
```
